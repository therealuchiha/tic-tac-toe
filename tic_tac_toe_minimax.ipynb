 Adversarial Search - Tic-Tac-Toe with Minimax
Introduction

I  implement an AI to play Tic-Tac-Toe optimally using the Minimax algorithm with alpha-beta pruning. Tic-Tac-Toe is a classic example of a deterministic, perfect-information, zero-sum game that can be solved completely using adversarial search algorithms.

This notebook covers:

The implementation of the core game logic
The Minimax algorithm implementation with alpha-beta pruning
A graphical user interface for human-AI interaction
Analysis of algorithm performance
1. Problem Formulation
Tic-Tac-Toe can be framed as a two-player deterministic adversarial game, where:

States: 3×3 game boards
Actions: Available positions (i, j) where a player can move
Players: Alternating between 'X' and 'O'
Terminal states: When the board is full or when a player has won
Utility: +1 for a win by X, -1 for a win by O, and 0 for a tie
2. Implementation
2.1 Game Logic

```python
"""Tic Tac Toe Game Logic"""

def initial_state():
    return [[None, None, None],
            [None, None, None],
            [None, None, None]]

def player(board):
    x_count = sum(row.count("X") for row in board)
    o_count = sum(row.count("O") for row in board)
    return "X" if x_count <= o_count else "O"

def actions(board):
    return {(i, j) for i in range(3) for j in range(3) if board[i][j] is None}

def result(board, action):
    new_board = [row[:] for row in board]
    i, j = action
    if new_board[i][j] is not None:
        raise Exception("Invalid action!")
    new_board[i][j] = player(board)
    return new_board

def winner(board):
    for row in board:
        if row[0] == row[1] == row[2] and row[0] is not None:
            return row[0]
    for j in range(3):
        if board[0][j] == board[1][j] == board[2][j] and board[0][j] is not None:
            return board[0][j]
    if board[0][0] == board[1][1] == board[2][2] and board[0][0] is not None:
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] and board[0][2] is not None:
        return board[0][2]
    return None

def terminal(board):
    if winner(board) is not None:
        return True
    return all(cell is not None for row in board for cell in row)

def utility(board):
    w = winner(board)
    return 1 if w == "X" else -1 if w == "O" else 0
```

2.2 Minimax with Alpha-Beta Pruning

```python
def minimax(board):
    return minimax_alpha_beta(board)

def minimax_alpha_beta(board):
    if terminal(board):
        return None

    current_player = player(board)

    if current_player == "X":
        best_val = float("-inf")
        best_action = None
        for action in actions(board):
            val = min_value(result(board, action), float("-inf"), float("inf"))
            if val > best_val:
                best_val = val
                best_action = action
        return best_action
    else:
        best_val = float("inf")
        best_action = None
        for action in actions(board):
            val = max_value(result(board, action), float("-inf"), float("inf"))
            if val < best_val:
                best_val = val
                best_action = action
        return best_action

def max_value(board, alpha, beta):
    if terminal(board):
        return utility(board)
    v = float("-inf")
    for action in actions(board):
        v = max(v, min_value(result(board, action), alpha, beta))
        if v >= beta:
            return v
        alpha = max(alpha, v)
    return v

def min_value(board, alpha, beta):
    if terminal(board):
        return utility(board)
    v = float("inf")
    for action in actions(board):
        v = min(v, max_value(result(board, action), alpha, beta))
        if v <= alpha:
            return v
        beta = min(beta, v)
    return v
```

3. Testing and Demonstration
3.1 Visualizing the Board

```python
def display_board(board):
    symbols = {None: " ", "X": "X", "O": "O"}
    print("-" * 13)
    for i, row in enumerate(board):
        print(f"| {symbols[row[0]]} | {symbols[row[1]]} | {symbols[row[2]]} |")
        if i < 2:
            print("-" * 13)
    print("-" * 13)
```

3.2 Testing Different Board States

```python
# Test Case 1
test_board1 = initial_state()
print("Test Case 1: Empty board")
display_board(test_board1)
print(f"AI's move: {minimax(test_board1)}")

# Test Case 2
test_board2 = [[None, None, None],
               [None, "X", None],
               [None, None, None]]
print("\nTest Case 2: One move in")
display_board(test_board2)
print(f"AI's move: {minimax(test_board2)}")

# Test Case 3
test_board3 = [["X", None, None],
               ["X", "O", None],
               [None, None, None]]
print("\nTest Case 3: AI needs to block")
display_board(test_board3)
print(f"AI's move: {minimax(test_board3)}")

# Test Case 4
test_board4 = [["X", "X", None],
               ["O", "O", None],
               ["X", None, None]]
print("\nTest Case 4: AI can win")
display_board(test_board4)
print(f"AI's move: {minimax(test_board4)}")
```

3.3 Simulating a Full Game

```python
def simulate_game():
    board = initial_state()
    print("Starting a new game simulation:")
    display_board(board)
    while not terminal(board):
        current = player(board)
        action = minimax(board)
        print(f"{current}'s move: {action}")
        board = result(board, action)
        display_board(board)
    result_winner = winner(board)
    if result_winner:
        print(f"Game over: {result_winner} wins!")
    else:
        print("Game over: It's a tie!")

simulate_game()
```
